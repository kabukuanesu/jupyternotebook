```python
# linear Search

def linear_search(items, target):
    for i in range(len(items)):
        if items[i] == target:
            return i
    return -1

items = [123, 9, 78, 54, 89, 5, 5, 7, 16]

print(linear_search(items, 54))
print(linear_search(items, 55))
```

```python
# Binary Search

def binary_search(items, target):
    low = 0
    high = len(items) - 1

    while low <= high:
        mid = (low + high) // 2
        if items[mid] == target:
            return mid
        elif items[mid] < target:
            low = mid + 1
        else:
            low = mid -1
    return -1

numbers = [1, 3, 5, 9, 10, 11, 18, 23, 30]

print(binary_search(numbers, 11))
print(binary_search(numbers, 20))
```
