## 4. Writing Functions

```python
# 4. Writing Functions
def total_marks(maths, physics, programming):
    return maths + physics + programming

df['Total'] = df.apply(lambda x: total_marks(x['Maths'], x['Physics'], x['Programming']), axis=1)
```

```python
df.columns
```

## 5. Average

```python
# 5. Average
def average_marks(total):
    return total / 3

df['Average'] = df.apply(lambda x: average_marks(x['Total']), axis=1)
```

```python
df.columns
```

## 6. Lambda Functions

```python
# 6. Lambda Again

df['Result'] = df['Average'].apply(lambda x: "Pass" if x >= 50 else "Fail")
```

```python
df.columns
```

## 7. Grade Calculation

```python
df['Grade'] = df['Average'].apply(
    lambda x: 'A' if x >= 90 else
              'B' if x >= 80 else
              'C' if x >= 70 else
              'D' if x >= 60 else
              'F')
```

```python
df[['Name', 'Total', 'Average', 'Result', 'Grade']]
```

## 9. Performance Analysis

```python
# 9. Performance Analysis

def department_report(df, dept):
    total = 0
    for i in df[df['Dept'] == dept]:
        total += 1
        print("Total:", total)
        return total

department_report(df, 'CSE')
```

## 13. Stack Implementation Using An Array

```python
# 13. Stack Implementation using Array

class Stack:
    def __init__(self):
        self.stack = []
        
    def push(self, item):
        self.stack.append(item)
        print(item, "pushed onto the stack")
        
    def pop(self):
        if len(self.stack) == 0:
            print("The Stack is empty")
        else:
            item = self.stack.pop()
            print(item, "popped from the stack")
            return item
            
    def display(self):
        if len(self.stack) == 0:
            print("The Stack is empty")
        else:
            print("Stack:", self.stack)

a = Stack()
a.display()
print()
a.push(13)
a.display()
print()
a.push(13)
a.display()
print()
a.push(78)
a.display()
print()
a.push(67)
a.display()
print()
a.pop()
a.display()
print()
a.pop()
a.display()
```
