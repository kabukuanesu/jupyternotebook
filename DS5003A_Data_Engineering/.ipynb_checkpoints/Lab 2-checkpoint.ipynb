## 8. Adding New Columns Using Functions

```python
def total_marks(maths, physics, programming):
    return maths + physics + programming

df["Total"] = df.apply(lambda row: total_marks(row["Maths"], row["Physics"], row["Programming"]), axis=1)
df[["Name","Total"]]
```

## 9. Average Marks

```python
def average_marks(total):
    return total/3

df["Average"] = df["Total"].apply(average_marks)
df[["Name","Average"]]
```

## 10. Lambda Functions

```python
df["Result"] = df["Average"].apply(lambda x: "Pass" if x>=50 else "Fail")
df[["Name","Average","Result"]]
```

```python
df["Grade"] = df["Average"].apply(
    lambda x: "A" if x>=90 else
              "B" if x>=80 else
              "C" if x>=70 else
              "D" if x>=60 else
              "F")
df[["Name","Average","Grade"]]
```

## 11. Creating a Dictionary

```python
roll_dict = dict(zip(df["Roll_No"], df["Name"]))
roll_dict
```

```python
roll_dict[101]
```

## 12. Basic Statistics

```python
print("Highest Programming Marks:", df["Programming"].max())
print("Lowest Programming Marks:", df["Programming"].min())
print("Average Programming Marks:", df["Programming"].mean())
```

## 13. Save the Updated DataFrame

```python
df.to_csv("students_updated.csv", index=False)
```
