## Graphs

```python
# Print a simple unweighted and undirected graph

graph = {
    "A": ["B", "C"],
    "B": ["D"],
    "C": ["E"],
    "D": [],
    "E": []
}

print(graph)
```

## Adding Nodes and Edges

```python
graph = {}

def add_node(graph, node):
    if node not in graph:
        graph[node] = []

def add_edge(graph, u, v):
    add_node(graph, u)
    add_node(graph, v)
    graph[u].append(v)
    graph[v].append(u)

add_edge(graph, "A", "B")
add_edge(graph, "A", "C")
add_edge(graph, "B", "D")
add_edge(graph, "C", "E")

print(graph)
```

## BFS

```python
# A Queue is used to implement BFS

from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)

    order = []

    while queue:
        node = queue.popleft()
        order.append(node)

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return order

graph = {
    "A": ["B", "C"],
    "B": ["B", "D"],
    "C": ["C", "E"],
    "D": [],
    "E": []
}

print("BFS:", bfs(graph, "A"))
```

## DFS Using Recursion

```python
def dfs(graph, node, visited=None):
    if visited is None:
        visited = set()

    visited.add(node)
    order = [node]

    for neighbor in graph[node]:
        if neighbor not in visited:
            order.extend(dfs(graph, neighbor, visited))

    return order

print("DFS:", dfs(graph, "A"))
```

## DFS Using An Explicit Stack

```python
def dfs_stack(graph, start):
    visited = set()
    stack = [start]
    order = []

    while stack:
        node = stack.pop()

        if node not in visited:
            visited.add(node)
            order.append(node)

            for neighbor in reversed(graph[node]):
                if neighbor not in visited:
                    stack.append(neighbor)

    return order

print(graph)

print("DFS:", dfs_stack(graph, "A"))
```

## Path Finding Using BFS

```python
from collections import deque

def bfs_path(graph, start, target):
    queue = deque([(start, [start])])
    visited = {start}

    while queue:
        node, path = queue.popleft()

        if node == target:
            return path

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [neighbor]))

    return None

graph = {
    "A": ["B", "C"],
    "B": ["D"],
    "C": ["E"],
    "D": [],
    "E": []
}

print("BFS Path:", bfs_path(graph, "A", "D"))
```

## Path Finding Using DFS

```python
def dfs_path(graph, start, target):
    stack = [(start, [start])]
    visited = set()

    while stack:
        node, path = stack.pop()

        if node in visited:
            continue

        visited.add(node)

        if node == target:
            return path

        for neighbor in reversed(graph[node]):
            if neighbor not in visited:
                stack.append((neighbor, path + [neighbor]))

    return None

print("DFS Path:", dfs_path(graph, "A", "D"))
```
