## Linear Search

```python
def linear_search(items, target):
    for i in range(len(items)):
        if items[i] == target:
            return i
    return -1

items = [12, 58, 14, 9, 78, 13, 6, 50]

print(linear_search(items, 9))
print(linear_search(items, 10))
```

## Binary Search

```python
def binary_search(items, target):
    low = 0
    high = len(items) - 1

    while low <= high:
        mid = (low + high) // 2
        if items[mid] == target:
            return mid
        elif items[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

items = [12, 58, 14, 9, 78, 13, 6, 50]
items.sort()
print(items)
print(binary_search(items, 6))
print(binary_search(items, 58))
print(binary_search(items, 11))
```

## Bubble Sort

```python
def bubble_sort(items):
    n = len(items)

    for i in range(n - 1):
        swapped = False

        for j in range(n - 1 - i):
            if items[j] > items[j + 1]:
                items[j], items[j + 1] = items[j + 1], items[j]
                swapped = True
        if not swapped:
            break
    return items

items = [12, 58, 14, 9, 78, 13, 6, 50]
print("Unsorted: ", items)
print("Sorted: ", bubble_sort(items))
```

## Selection Sort

```python
def selection_sort(items):
    n = len(items)

    for i in range(n - 1):
        min_index = i

        for j in range(i + 1, n):
            if items[j] < items[min_index]:
                min_index = j

        items[i], items[min_index] = items[min_index], items[i]

    return items

items = [12, 58, 14, 9, 78, 13, 6, 50]
print("Unsorted:", items)
print("Sorted:", selection_sort(items))
```

## Insertion Sort

```python
def insertion_sort(items):
    for i in range(1, len(items)):
        current = items[i]
        j = i - 1

        while j >= 0 and items[j] > current:
            items[j + 1] = items[j]
            j = j - 1

        items[j + 1] = current

    return items

items = [12, 58, 14, 9, 78, 13, 6, 50]
print("Unsorted:", items)
print("Sorted:", insertion_sort(items))
```

## Merge Sort

```python
def merge_sort(items):

    if len(items) <= 1:
        return items

    mid = len(items) // 2

    left = items[:mid]
    right = items[mid:]

    left = merge_sort(left)
    right = merge_sort(right)

    result = []

    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    while i < len(left):
        result.append(left[i])
        i += 1

    while j < len(right):
        result.append(right[j])
        j += 1

    return result

items = [12, 58, 14, 9, 78, 13, 6, 50]
print("Unsorted:", items)
print("Sorted:", merge_sort(items))
```

## Quick Sort

```python
def quick_sort(items):

    # Base case
    if len(items) <= 1:
        return items

    # Choose the last element as the pivot
    pivot = items[-1]

    # Lists for elements smaller and greater than pivot
    left = []
    right = []

    # Compare each element with the pivot
    for item in items[:-1]:
        if item <= pivot:
            left.append(item)
        else:
            right.append(item)

    # Recursively sort left and right
    return quick_sort(left) + [pivot] + quick_sort(right)


# Main program
numbers = [8, 3, 5, 4, 7, 6, 1, 2]

print("Original list:", numbers)

sorted_numbers = quick_sort(numbers)

print("Sorted list:", sorted_numbers)
```
