## 1. Linear Search

Linear search checks each element of a list one by one until the required element is found.

It can be used on both **sorted and unsorted lists**.

### How Linear Search Works

Linear Search starts from the **first element** and checks elements one at a time until the target is found or the list ends.

Example:

`[23, 12, 45, 7, 19]` &nbsp;&nbsp; Target = `7`

1. Check `23` → not the target.
2. Check `12` → not the target.
3. Check `45` → not the target.
4. Check `7` → **target found**.



A useful way to remember Linear Search is:

**Start from the beginning → Check one by one → Stop when found**

Linear Search does **not** require the list to be sorted.


```python
numbers = [24, 10, 35, 18, 42, 7]
target = 18

for i in range(len(numbers)):
    if numbers[i] == target:
        print("Element found at index:", i)
        break
else:
    print("Element not found")
```

### Linear Search Using a Function

The function returns the index of the element if it is found.  
If the element is not present, it returns `-1`.

```python
def linear_search(items, target):
    for i in range(len(items)):
        if items[i] == target:
            return i
    return -1
```

```python
numbers = [15, 8, 21, 30, 12]

print(linear_search(numbers, 21))
print(linear_search(numbers, 50))
```

## 2. Binary Search

Binary search repeatedly divides the search range into two halves.

**Important:** Binary search works only when the list is **sorted**.

### How Binary Search Works

Binary Search works by repeatedly checking the **middle element** of a sorted list.

Example:

`[3, 8, 12, 17, 23, 31, 42]` &nbsp;&nbsp; Target = `31`

1. Check the middle value `17`.
2. Since `31 > 17`, the target can only be in the **right half**.
3. Ignore the left half and check the middle of the remaining part.
4. The next middle value is `31` → **target found**.



A useful way to remember Binary Search is:

**Check middle → Choose left or right half → Repeat**

> **Important:** Binary Search requires the data to be **sorted**.


```python
numbers = [5, 10, 15, 20, 25, 30, 35]
target = 25

low = 0
high = len(numbers) - 1

while low <= high:
    mid = (low + high) // 2

    if numbers[mid] == target:
        print("Element found at index:", mid)
        break
    elif numbers[mid] < target:
        low = mid + 1
    else:
        high = mid - 1
else:
    print("Element not found")
```

### Binary Search Using a Function

```python
def binary_search(items, target):
    low = 0
    high = len(items) - 1

    while low <= high:
        mid = (low + high) // 2

        if items[mid] == target:
            return mid
        elif items[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1
```

```python
numbers = [4, 9, 13, 18, 25, 31, 40]

print(binary_search(numbers, 18))
print(binary_search(numbers, 27))
```

If the original list is not sorted, sort it before applying binary search.

```python
numbers = [32, 11, 45, 7, 19, 26]

numbers.sort()
print("Sorted list:", numbers)

print("Index of 19:", binary_search(numbers, 19))
```

## 3. Bubble Sort

Bubble sort repeatedly compares two adjacent elements and swaps them if they are in the wrong order.

After each pass, the largest remaining element moves towards the end of the list.

### How Bubble Sort Works

Bubble Sort compares **adjacent elements** from left to right.

If the left element is greater than the right element, the two elements are swapped.  
After one complete pass, the **largest unsorted element reaches its correct position at the end**.

For example, consider:

`[5, 1, 4, 2, 8]`

- Compare `5` and `1` → swap them.
- Compare `5` and `4` → swap them.
- Compare `5` and `2` → swap them.
- Compare `5` and `8` → no swap.
- Repeat the process for the remaining unsorted part.



A useful way to remember Bubble Sort is:

**Compare neighbours → Swap if needed → Repeat passes**


```python
numbers = [5, 1, 4, 2, 8]

print("Before sorting:", numbers)

for i in range(len(numbers) - 1):
    if numbers[i] > numbers[i + 1]:
        numbers[i], numbers[i + 1] = numbers[i + 1], numbers[i]

print("After one pass:", numbers)
```

### Bubble Sort Using a Function

```python
def bubble_sort(items):
    n = len(items)

    for i in range(n - 1):
        swapped = False

        for j in range(n - 1 - i):
            if items[j] > items[j + 1]:
                items[j], items[j + 1] = items[j + 1], items[j]
                swapped = True

        if not swapped:
            break

    return items
```

```python
numbers = [64, 34, 25, 12, 22, 11, 90]

print("Before sorting:", numbers)
print("After sorting :", bubble_sort(numbers))
```

## 4. Selection Sort

Selection sort finds the smallest element from the unsorted portion of the list and places it in the correct position.

### How Selection Sort Works

Selection Sort divides the list into two parts:

- a **sorted portion** on the left, and
- an **unsorted portion** on the right.

For each position, it searches the unsorted portion for the **smallest element** and places that element in the current position.

For example:

`[29, 10, 14, 37, 13]`

1. The smallest value is `10`, so place it first.
2. From the remaining values, the smallest is `13`, so place it second.
3. Continue until the complete list is sorted.



A useful way to remember Selection Sort is:

**Find minimum → Put it in the next position → Repeat**


```python
def selection_sort(items):
    n = len(items)

    for i in range(n - 1):
        min_index = i

        for j in range(i + 1, n):
            if items[j] < items[min_index]:
                min_index = j

        items[i], items[min_index] = items[min_index], items[i]

    return items
```

```python
numbers = [29, 10, 14, 37, 13]

print("Before sorting:", numbers)
print("After sorting :", selection_sort(numbers))
```

## 5. Insertion Sort

Insertion sort takes one element at a time and inserts it into the correct position in the already sorted part of the list.

### How Insertion Sort Works

Insertion Sort builds the sorted list **one element at a time**.

Think of arranging playing cards in your hand: take the next card and insert it into the correct position among the cards that are already arranged.

For example:

`[12, 11, 13, 5, 6]`

- Start with `12` as the sorted portion.
- Insert `11` before `12`.
- `13` is already in the correct place.
- Insert `5` before `11`.
- Insert `6` between `5` and `11`.



A useful way to remember Insertion Sort is:

**Take next element → Shift larger elements → Insert in the correct place**


```python
def insertion_sort(items):
    for i in range(1, len(items)):
        current = items[i]
        j = i - 1

        while j >= 0 and items[j] > current:
            items[j + 1] = items[j]
            j = j - 1

        items[j + 1] = current

    return items
```

```python
numbers = [12, 11, 13, 5, 6]

print("Before sorting:", numbers)
print("After sorting :", insertion_sort(numbers))
```

## 6. Merge Sort — Concept

Merge Sort follows a **divide-and-conquer** approach.

Instead of sorting the whole list at once, it repeatedly divides the list into smaller parts. Once the parts become very small, they are combined back together in sorted order.

For example:

`[8, 3, 5, 4, 7, 6, 1, 2]`

### Main idea

1. **Divide** the list into two halves.
2. Keep dividing each half until individual elements or very small lists remain.
3. **Merge** neighbouring parts while arranging their elements in sorted order.
4. Continue merging until one completely sorted list remains.



A useful way to remember Merge Sort is:

**Divide → Divide again → Merge in sorted order**

### Key point

During the merge step, two already sorted portions are compared and combined into one larger sorted portion.

Merge Sort has a typical time complexity of **O(n log n)**.


## 7. Quick Sort — Concept

Quick Sort is also based on the **divide-and-conquer** idea, but it divides the list using a special element called a **pivot**.

For example:

`[9, 4, 7, 3, 10, 5]`

Suppose `5` is chosen as the pivot.

### Main idea

1. Choose a **pivot** element.
2. Move values **smaller than the pivot** to one side.
3. Move values **larger than the pivot** to the other side.
4. The pivot is now in its appropriate position relative to those values.
5. Repeat the same process on the left and right portions.



A useful way to remember Quick Sort is:

**Choose pivot → Partition → Repeat on smaller parts**

### Key point

The exact intermediate arrangement can vary depending on the pivot and partition method used, but the central idea remains the same.

Quick Sort has an average time complexity of **O(n log n)**, although its worst case is **O(n²)**.


## 8. Python Built-in Sorting

Python provides two commonly used ways to sort a list:

- `list.sort()` changes the original list.
- `sorted()` returns a new sorted list without changing the original list.


```python
numbers = [30, 10, 50, 20, 40]

numbers.sort()
print(numbers)
```

```python
numbers = [30, 10, 50, 20, 40]

sorted_numbers = sorted(numbers)

print("Original list:", numbers)
print("Sorted list  :", sorted_numbers)
```

### Sorting in Descending Order

Use `reverse=True` to sort from largest to smallest.

```python
numbers = [30, 10, 50, 20, 40]

numbers.sort(reverse=True)
print(numbers)
```

### Sorting Strings

```python
names = ["Ravi", "Anu", "Kiran", "Binu", "Dev"]

names.sort()
print(names)
```

### Sorting Using `key`

The `key` parameter can be used to decide which value should be used for sorting.

The following example sorts student records using marks.

```python
students = [
    ("Anu", 78),
    ("Ravi", 91),
    ("Meera", 84),
    ("Kiran", 69)
]

students.sort(key=lambda student: student[1])

print(students)
```

```python
students.sort(key=lambda student: student[1], reverse=True)

print(students)
```

## 9. Searching Using Python List Operations

For simple programs, Python also provides built-in operations for checking and locating an item.


```python
numbers = [10, 20, 30, 40, 50]

print(30 in numbers)
print(70 in numbers)
```

```python
numbers = [10, 20, 30, 40, 50]

print(numbers.index(30))
```

`index()` gives an error if the item is not present.  
We can check using `in` before calling `index()`.

```python
target = 70

if target in numbers:
    print("Index:", numbers.index(target))
else:
    print("Element not found")
```

## 10. Time Complexity

| Algorithm | Time Complexity | Requirement / Note |
|---|---:|---|
| Linear Search | O(n) | Works on sorted or unsorted lists |
| Binary Search | O(log n) | List must be sorted |
| Bubble Sort | O(n²) | Simple comparison-based sorting |
| Selection Sort | O(n²) | Selects the minimum element in each pass |
| Insertion Sort | O(n²) | Useful for small or nearly sorted lists |
| Merge Sort | O(n log n) | Divide-and-conquer; merges sorted portions |
| Quick Sort | O(n log n) average | Uses a pivot; worst case O(n²) |
| Python `sort()` / `sorted()` | O(n log n) | Preferred for normal Python programs |


## Summary

- **Linear Search** checks elements one by one.
- **Binary Search** repeatedly halves the search space and requires sorted data.
- **Bubble Sort** swaps adjacent elements.
- **Selection Sort** repeatedly selects the smallest element.
- **Insertion Sort** inserts each element into its correct position.
- **Merge Sort** divides the list and merges sorted portions.
- **Quick Sort** chooses a pivot and partitions values around it.
- For regular Python programs, `sort()` and `sorted()` are usually preferred.
