```python
# 1. Temperature Conversion

celcius = float(input("Enter Temperature in Celcius:"))

fahrenheit = (9/5) * celcius + 32

print("Temperature in Fahrenheit:", fahrenheit)
```

```python
# 2. Simple Interest

principal = float(input("Enter the Principal Amount:"))
rate = float(input("Enter the rate (%):"))
time = float(input("Enter the time in years:"))

interest = (principal * rate * time) / 100
total = principal + interest

print("Simple Interest:", interest)
print("Total Amount:", total)
```

```python
# 3. Largest Number

num1 = float(input("Enter the first number:"))
num2 = float(input("Enter the second number:"))
num3 = float(input("Enter the third number:"))

if (num1 > num2) and (num1 > num3):
    print("Largest Number:", num1)
elif (num2 > num1) and (num2 > num3):
    print("Largest Number:", num2)
elif (num3 > num1) and (num3 > num2):
    print("Largest Number:", num3)
else:
    print("There are equal numbers present")
```

```python
# 4. Sides of a triangle

side1 = float(input("Enter side 1:"))
side2 = float(input("Enter side 2:"))
side3 = float(input("Enter side 3:"))

if (side1 < side2 + side3) and (side2 < side1 + side3) and (side3 < side1 + side2):
    print("The three sides are of a triangle")
else:
    print("The sides are not of a triangle")
    
```

```python
# 5. Palindrome

number = int(input("Enter a number:"))

original = number
reverse = 0

while (number > 0):
    digit = number % 10
    reverse = reverse * 10 + digit
    number = number // 10
    
print(f"Reverse: {reverse}")
print("")

if (reverse == original):
    print("The number is a palindrome")
else:
    print("The number is not a palindrome")
```

```python
# 6. Sum of Digits

number = int(input("Enter a number:"))
sum_of_digits = 0

while number > 0:
    digit = number % 10
    sum_of_digits += digit
    number = number // 10

print("Sum of digits:", sum_of_digits)
```

```python
# 7. Explicit Casting

name = str(input("Enter your name:"))
age = str(input("Age:"))

# age = int(age)
# age += 1
# age = str(age)

def next_age(age):
    age = int(age)
    age += 1
    age = str(age)
    return age

age = next_age(age)

print("Hello " + name + ", you will be " + age + " years old next year.")
```

```python
# 8. Euclidean Algorithm

num1 = int(input("Enter the first number:"))
num2 = int(input("Enter the second number:"))

a = num1
b = num2

while b != 0:
    temp = b
    b = a % b
    a = temp

print("GCD:", a)
```

```python
# 11. Star Pyramid

H = int(input("Enter a number:"))

for i in range(1, H + 1):

    for j in range(H-i):
        print(" ", end="")

    for k in range(2 * i - 1):
        print("*", end="")

    print()
```

```python
# 12. Lists

n = int(input("Enter the number of elements:"))
numbers = []
largest = 0
total = 0

for i in range(n):
    num = int(input("Enter element " + str(i + 1) + ":"))
    numbers.append(num)

print("List Elements: ", numbers)

for num in numbers:
    total += num

    if num > largest:
        largest = num

print("Sum of Elements: ", total)
print("Largest Element: ", largest)
```

```python
# 13. Perfect Divisor

def is_perfect(n):
    total = 0
    for i in range(1, n):
        if n % i == 0:
            total += i
    return total == n

limit = int(input("Enter upper limit:"))
print("Perfect numbers between 1 and", str(limit), ": ", end="")

for i in range(1, limit + 1):
    if is_perfect(i):
        print(i, end=" ")
```
