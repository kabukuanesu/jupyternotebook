```python
# 1. Celcius to Fahreinheit

celcius = float(input("Enter temperature in Celcius: "))

fahreinheit = (9 / 5) * celcius + 32

print("Temperature in Fahreinheit: ", fahreinheit)
```

```python
# 2. Simple Interest and Total Amount

principal = float(input("Enter the principal amount: "))
rate = float(input("Enter the rate of interest (%): "))
time = float(input("Enter the time (in years): "))

simple_interest = (principal * rate * time) / 100
total_amount = principal + simple_interest

print("Simple Interest: ", simple_interest)
print("Total Amount: ", total_amount)
```

```python
# 3. Largest of three numbers

num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
num3 = float(input("Enter the third number: "))

if num1 >= num2 and num1 >= num3:
    largest = num1
elif num2 >= num1 and num2 >= num3:
    largest = num2
else:
    largest = num3

print("Largest number: ", largest)
```

```python
# 4. If three sides can form a triangle

side1 = float(input("Enter the first side: "))
side2 = float(input("Enter the second side: "))
side3 = float(input("Enter the third side: "))

if (side1 + side2 > side3) and (side1 + side3 > side2) and (side2 + side3 > side1):
    print("The given sides can form a triangle.")
else:
    print("The given sides cannot form a triangle.")
    
```

```python
# 5. Reverse a number and check if it is a palindrome

number = int(input("Enter a number: "))

original = number
reverse = 0

while number > 0:
    digit = number % 10 # get last digit
    reverse = reverse * 10 + digit
    number = number // 10 # remove the last digit

print("Reversed number: ", reverse)

if original == reverse:
    print("The number is a palindrome")
else:
    print("The number is not a palindrome")
    
```

```python
# 6. Sum of the digits

number = int(input("Enter a number: "))

sum_digits = 0

while number > 0:
    digit = number % 10
    sum_digits += digit
    number = number // 10

print("Sum of digits: ", sum_digits)
```

```python
# 7. Displaying name and age

name = input("Enter your name: ")
age = input("Enter your age: ")

age = int(age)
age = age + 1
age = str(age)

print("Hello " + name + ", you will be " + age + " years old next year.")
```

```python
# 8. The Greatest Common Divisor (GCD)

num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

a = num1
b = num2

while b != 0:
    temp = b
    b = a % b
    a = temp

print("GCD:", a)
```

```python
# 9. The multiplication table of a number

n = int(input("Enter the number: "))
m = int(input("Enter the limit: "))

for i in range(1, m + 1):
    print(n, "x", i, "=", n * i)
    
```

```python
# 10. Number pattern

n = int(input("Enter the number of rows: "))

for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
    
```

```python
# 11. A centered star pyramid

H = int(input("Enter the height of the pyramid: "))

for i in range(1, H + 1):
    # Print leading spaces
    for j in range(H - i):
        print(" ", end="")
    
    # Print stars
    for k in range(2 * i - 1):
        print("*", end="")
    
    # Move to the next line
    print()
    
```

```python
# 12. Read n integers, display the list, calculate the sum, and find the largest element

n = int(input("Enter the number of elements: "))

numbers = []

for i in range(n):
    num = int(input("Enter element " + str(i + 1) + ": "))
    numbers.append(num)

print("List elements:", numbers)

total = 0
for num in numbers:
    total += num

largest = numbers[0]
for num in numbers:
    if num > largest:
        largest = num

print("Sum of elements:", total)
print("Largest element:", largest)
```

```python
# 13. Function to check whether a number is perfect

def is_perfect(n):
    total = 0

    for i in range(1, n):
        if n % i == 0:
            total += i

    return total == n


limit = int(input("Enter the upper limit: "))

print("Perfect numbers between 1 and", limit, ":", end=" ")

for i in range(1, limit + 1):
    if is_perfect(i):
        print(i, end=" ")
        
```
