### IF Statement

### IF-ELSE Statement
* We can use the else statement with if statement to execute a block of code when the condition is false.

```python
# Python program to wish Good Morning or Good Evening based on time.
import time

time_tuple = time.localtime() # get struct_time
hour=time_tuple.tm_hour
if(hour >= 0 and hour < 12): # Checking condition
    print("Good Morning")
else:
    print("Good Evening")
```

### Nested IF Statement
* If statement inside the if-block of another if statement and so on - this is called a nested if statement.

```python
# Write a programe to user login.
user_name="John"
user_password="programming"
input_user_name=input("Enter User Name : ")
if(input_user_name == user_name):      # outer if statement
    input_user_password=input("Enter User Password : ")
    if(user_password == input_user_password):   # inner if statement
        print("Login Sucessful")
    else:
        print("Incorrect Password")
else:
    print("User not found")
```

## Loops
* Loops in programming come into use when we need to repeatedly execute a block of statements.


* WHILE
    * The while loop allows you to repeatedly execute a block of statements as long as a condition is true. 

* FOR
    * The for loop is another looping statement which iterates over a sequence items and executes a block of code once for each item in the sequence.

### Nested loops
* Executing loop inside a loop.

### Loop Control Statements
* Break
    * The break statement is used to break out of a loop statement i.e. stop the execution of a looping statement, even if the loop condition has not become False or the sequence of items has not been completely iterated over.
    
* Continue
    * he continue statement is used to tell Python to skip the rest of the statements in the current loop block and to continue to the next iteration of the loop.



```python
# Python program to print tables using for loop.
n=int(input("Enter Table Number : "))

print(f"{n}th Table :")  #f-string
for i in range(1,11):
    print(f"{n}*{i}={n*i}")
    i+=1
```

```python
# Python programe for checking login credentials. Check login credintials untill correct values were given.
user_name="John"
user_password="programming"
while(True):
    input_user_name=input("Enter user name : ")
    if(input_user_name == user_name):
        input_user_password=input("Enter user password : ")
        if(user_password == input_user_password):
            print("Login Sucessful")
            break   #break the loop
        else:
            print("Sorry incorrect password, Try again")
    else:
        print("Sorry user not found, Try again")
```

```python
'''
Python program to construct the following pattern, using a nested for loop.

* 
* * 
* * * 
* * * * 
* * * * * 
* * * * 
* * * 
* * 
*
'''

n=5;
for i in range(n):    # Outer for loop
    for j in range(i):   # Inner for loop
        print ('* ', end="")
    print(" ")

for i in range(n,0,-1):
    for j in range(i):
        print('* ', end="")
    print(" ")
```

## Functions

* Functions are reusable pieces of programs. They allow you to give a name to a block of statements, allowing you to run that block using the specified name anywhere in your program and any number of times. This is known as calling the function.


* Functions are defined using the def keyword. After this keyword comes an identifier name for the function, followed by a pair of parentheses which may enclose some names of variables, and by the final colon that ends the line. Next follows the block of statements that are part of this function.


#### Defining and Calling a function

```python
# Defining Function
def welcome():
    print("Hi, Welcome to Functions")
    
# Calling function
welcome()  
```

### Function Parameters
* A function can take parameters, which are values you supply to the function so that the function can do something utilising those values.

* Parameters are specified within the pair of parentheses in the function definition, separated by commas.


```python
# Parameter name is passing to welcome function.
def welcome(name):
    print("Hi "+name+", Welcome to Functions.")

welcome("Albert")
welcome("Mike")
```


### Default Argument Values
* You can specify default argument values for parameters by appending to the parameter name in the function definition

```python
# Here Default argument of parameter name is John.
def welcome(name="John"):
    print("Hi "+name+", Welcome to Functions.")

welcome("Mike")
welcome()
```

### Return Statement
* Functions whose primary job is to display information and other functions designed to process data and return a value or set of values.
* The return statement takes a value from inside a function and sends it back to the line that called the function.

```python
# Programe to print lower case of a given string.
def LowerCaseConversion(string):
    lower_string=string.lower()
    return lower_string

text="HellO"
lower_text=LowerCaseConversion(text)
print("Lower case of given string is : "+lower_text)
```

### Local and Global Variables
* Scope of variable
* A variable declared inside the function's body or in the local scope is known as a local variable.
* Global variables are the one that is defined and declared outside a function. It have global scope.
* If a variable with the same name is defined inside the scope of function as well then it will print the value given inside the function only and not the global value.
* Need to use global keyword to modify global variable inside the function

```python
def foo():
    print(string1)
    string2="Hi I'm local variable"
    print(string2)

string1="Hi I'm global variable"
foo()
```

```python
# Modifying the global variable in the function
def foo():
    global string1
    print(string1)
    string1="Hi I'm local variable"
    print(string1)

string1="Hi I'm global variable"
foo()
print(string1)
```

```python
# Python program to print the all prime number between given two numbers.
def prime_function(n):
    if(n==1 or n==0):   # Checking base condition.
        return False
    else:
        for p in range(2,n):  # for loop 
            if(n%p == 0):
                return False
            else:
                continue
        return True
num1,num2=input("Enter the range of numbers separated by space:").split() # split function.
num1=int(num1)  #type casting.
num2=int(num2)
print(f"Prime number between {num1} and {num2} : ")
for i in range(num1,num2+1):
    if(prime_function(i)):
        print(i,end=" ")
```

### Lambda Function
* Anonymous functions also are called Lambda functions.Anonymous function means that a function is without a name.The lambda keyword is used to define an anonymous function in Python. 
* <b> Syntax: lambda arguments: expression </b>
* This function can have any number of arguments but only one expression, which is evaluated and returned.

```python
# Python program to illustrate cube of a number showing difference between def() and lambda().
def cube(y):
    return y*y*y

# defining lambda function for getting cube of a number
lambda_cube = lambda y: y*y*y
 
# using the normally defined function
print(cube(5))
 
# using the lambda function
print(lambda_cube(5))
```

```python
# lambda function for addition of two numbers
addition = lambda x,y : x+y
print(addition(10,20))
```
