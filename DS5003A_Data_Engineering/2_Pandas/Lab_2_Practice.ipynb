## 11. Student Class

```python
class Student:
    
    def __init__(self, roll_no, name, dept):
        self.roll_no = roll_no
        self.name = name
        self.dept = dept
        self.maths = 0
        self.physics = 0
        self.programming = 0
        self.attendance = 0
        
    def set_maths(self, marks):
        self.maths = marks
        
    def set_physics(self, marks):
        self.physics = marks
        
    def set_programming(self, marks):
        self.programming = marks
        
    def set_attendance(self, attendance):
        self.attendance = attendance

    def average_marks(self):
        return (self.maths + self.physics + self.programming) / 3
        
    def scholarship_eligible(self):
        return self.average_marks() >= 85 and self.attendance >= 90
            
    def performance_warning(self):
        if (self.maths < 30 and
            self.physics < 30 and
            self.programming < 30):
            return "Warning: Poor Performance In All Courses!"
        else:
            return "Satisfactory"

    def __str__(self):
        return(
            f"Roll No: {self.roll_no}, "
            f"Name: {self.name}, "
            f"Department: {self.dept}, "
            f"Average: {self.average_marks():.2f} "
        )

students = []

for _, row in df.iterrows():
    student = Student(row['Roll_No'], row['Name'], row['Dept'])

    student.set_maths(row['Maths'])
    student.set_physics(row['Physics'])
    student.set_programming(row['Programming'])
    student.set_attendance(row['Attendance'])

    students.append(student)

for student in students:
    print(student)
    print("Scholarship Eligible: ", student.scholarship_eligible())
    print(student.performance_warning())
    print("-" * 40)
```

## 13. Implementation of a Stack using an Array

```python
class Stack:
    def __init__(self):
        self.stack = []
        
    def push(self, item):
        self.stack.append(item)
        print(f"{item} pushed into stack")
        
    def pop(self):
        if len(self.stack) == 0:
            print("Stack Underflow")
        else:
            item = self.stack.pop()
            print(f"{item} popped out of the stack")
            
    def display(self):
        print("Stack: ", self.stack)

stack = Stack()
stack.display()
stack.pop()
stack.push(13)
stack.display()
stack.push(13)
stack.display()
stack.push(78)
stack.display()
stack.push(67)
stack.display()
stack.pop()
stack.display()
stack.pop()
stack.display()
```

## 14. Implementation of a Stack using a Linked-List

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class Stack:
    def __init__(self):
        self.top = None

    def push(self, data):
        new_node = Node(data)
        new_node.next = self.top
        self.top = new_node

    def pop(self):
        if self.top == None:
            print('Stack Underflow')
        else:
            popped = self.top.data
            print(f"{popped} popped out of the stack")
            self.top = self.top.next

    def display(self):
        if self.top == None:
            print('Stack Underflow')
        else:
            current = self.top
            print("Stack: ", end="")
            while current:
                print(current.data, end=" ")
                current = current.next
            print()

s = Stack()
s.display()
s.push(13)
s.display()
s.push(13)
s.display()
s.push(78)
s.display()
s.push(67)
s.display()
s.pop()
s.display()
s.pop()
s.display()
```
