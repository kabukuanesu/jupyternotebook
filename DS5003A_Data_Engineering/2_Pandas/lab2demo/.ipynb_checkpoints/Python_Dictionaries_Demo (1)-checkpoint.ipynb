## Python Dictionaries
- Storing multiple items in key, value pairs
- Created using curly brackets
- Mutable
- Ordered
- Duplicate keys not allowed.
- Key can be any primitive data types
- Value can be any data type

```python
scores = {'Ram': 48, 'John':25, 'Athul': 40}
print(scores)
```

What happens if keys are duplicates??

```python
scores = {'Ram': 48, 'John':25, 'Athul': 40, 'Ram':30}
print(scores)
```

Value can be any data type

```python
myDict = {'A': 10, 'B':[1,2,3], 'C':'Hello'}
print(myDict)
```

Creating an empty dictionary with dict()

```python
empty_dict = dict()
print(empty_dict)
```

Length of a dictionary can be calculated with len() function

```python
print(len(scores))
```

Using type() function we can verify the given data structure is a dictionary or not

```python
print(type(scores))
```

### Accessing Dictionary Items

1. by reffering to its key name inside a square bracket

```python
scores = {'Ram': 48, 'John':25, 'Athul': 40}
print(scores["Ram"])
```

2. using *get()* method

```python
print(scores.get("Ram"))
```

### Get Keys,values in the dictionary
keys() method will return list of keys in the dictionary

```python
keysList = scores.keys()
print(keysList)
```

values() method will return list of values in the dictionary

```python
keysValues = scores.values()
print(keysValues)
```

items() gives the list of key value pairs in the dictionary

```python
items = scores.items()
print(items)
```

```python
# print all key value pairs with loop
for key,value in scores.items():
  print(key, '\t', value)
```

### Changing Values

1. by reffering to key name

```python
scores = {'Ram': 48, 'John':25, 'Athul': 40}
scores['John'] = 50
print(scores)
```

2. with update() method

```python
scores.update({'Ram':10})
print(scores)
```

```python
scores.update({'Ram':25,'John':21}) # update two values at a time
print(scores)
```

```python
# arguments no need to be a dictionary
# we can give a list of key value pairs
scores.update([('Ram',11),('John',10)])
print(scores)
```

```python
# tuple of key value pairs
scores.update((('Ram',50),('John',50)))
print(scores)
```

### Adding New Items

1. by using a new index key and assigning a value to it

```python
scores = {'Ram': 48, 'John':25, 'Athul': 40}
scores['Shruthi'] = 45
print(scores)
```

2. Using update() method. If the item does not exist, the item will be added else it will update

```python
scores = {'Ram': 48, 'John':25, 'Athul': 40}
scores.update({'Shruthi':21})
print(scores)
```

1. remove an item with specified key

---


- *dict.pop(index)* : pop() function return removed item

```python
scores = {'Ram': 48, 'John':25, 'Athul': 40}
item = scores.pop('Ram')
print(item)          # print removed value
print(scores)        # print dictionary after removal
```

- popitem() remove last inserted key value pair : it return removed key value pair

```python
scores = {'Ram': 48, 'John':25, 'Athul': 40}
item = scores.popitem()
print(item)
print(scores)
```

- *del list[index]*

```python
scores = {'Ram': 48, 'John':25, 'Athul': 40}
del scores['Ram']
print(scores)
```

 - del dictionary : delete entire dictionary
 - dictionary.clear()  : empty entire dictionary

### Loop Through a dictionary

When looping through a dictionary using for, the return value are the keys of the dictionary

```python
scores = {'Ram': 48, 'John':25, 'Athul': 40}
for x in scores:
  print(x)
```

```python
# print all values
for x in scores:
  print(scores[x])
```

```python
# another way of printing values
for x in scores.values():
  print(x)
```

```python
# another way of printing keys
for x in scores.keys():
  print(x)
```

```python
# print both keys and values
for x,y in scores.items():
  print(x,y)
```

### Dictionary Comprehension

1.  Make dictionary from a list using comprehension

```python
myDict = {x: x**2 for x in [1,2,3,4,5]}
print (myDict)
```

2. Dictionary comprehensions with if and else statements

```python
newdict = {x: x**3 for x in range(10) if x**3 % 4 == 0}
print(newdict)
```
