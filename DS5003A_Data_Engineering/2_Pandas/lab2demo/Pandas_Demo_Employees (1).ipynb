## Functions

```python
def annual_salary(monthly_salary):
    return monthly_salary * 12

df["Annual_Salary"] = df["Salary"].apply(annual_salary)
df[["Employee","Salary","Annual_Salary"]]
```

## Lambda Function

```python
df["Bonus_Eligible"] = df["Performance"].apply(
    lambda x: "Yes" if x=="Excellent" else "No"
)
df[["Employee","Performance","Bonus_Eligible"]]
```

## Dictionary

```python
emp_dict = dict(zip(df["Emp_ID"], df["Employee"]))
emp_dict
```

## Class Example

```python
class EmployeeReport:
    def annual_salary(self, salary):
        return salary * 12

report = EmployeeReport()
df["Annual_Salary"] = df["Salary"].apply(report.annual_salary)
df[["Employee","Annual_Salary"]]
```
