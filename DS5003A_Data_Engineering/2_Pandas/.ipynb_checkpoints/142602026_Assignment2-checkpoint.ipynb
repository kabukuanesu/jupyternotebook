## 4. Functions

```python
def total_marks(maths, physics, programming):
    return maths + physics + programming

df['Total'] = df.apply(
    lambda row: total_marks(row['Maths'], row['Physics'], row['Programming']),
    axis=1
)

df
```

## 5. Average

```python
def average_marks(total):
    return total / 3

df['Average'] = df['Total'].apply(average_marks)

df
```

## 6. Lambda Functions

```python
df['Result'] = df['Average'].apply(lambda x: 'Pass' if x >= 50 else 'Fail')

df
```

## 7. Grade Calculation

```python
df['Grade'] = df['Average'].apply(
    lambda x: 'A' if 90 <= x <= 100
    else 'B' if 80 <= x < 90
    else 'C' if 70 <= x < 80
    else 'D' if 60 <= x < 70
    else 'F'
)

df
```

## 9. Department Performance Analysis

```python
def department_report(df, dept):
    dept_df = df[df['Dept'] == dept]

    return {
        "Total Students": len(dept_df),
        "Average Programming Marks": dept_df['Programming'].mean(),
        "Highest Programming Marks": dept_df['Programming'].max()
    }

department_report(df, "CSE")
```

## 13. Student Class

```python
import pandas as pd

class Student:
    def __init__(self, roll_no, name, dept):
        self.roll_no = roll_no
        self.name = name
        self.dept = dept
        self.maths = 0
        self.physics = 0
        self.programming = 0
        self.attendance = 0

    # Functions to update marks and attendance
    def set_maths(self, marks):
        self.maths = marks

    def set_physics(self, marks):
        self.physics = marks

    def set_programming(self, marks):
        self.programming = marks

    def set_attendance(self, attendance):
        self.attendance = attendance

    # Calculate average marks
    def average_marks(self):
        return (self.maths + self.physics + self.programming) / 3

    # Check scholarship eligibility
    def scholarship_eligible(self):
        return self.average_marks() >= 85 and self.attendance >= 90

    # Check performance and issue warning
    def performance_warning(self):
        if (self.maths < 30 and
            self.physics < 30 and
            self.programming < 30):
            return "Warning: Poor performance in all courses."
        else:
            return "Performance is satisfactory."

    def __str__(self):
        return (f"Roll No: {self.roll_no}, "
                f"Name: {self.name}, "
                f"Dept: {self.dept}, "
                f"Average: {self.average_marks():.2f}")

students = []

for _, row in df.iterrows():
    student = Student(row['Roll_No'], row['Name'], row['Dept'])

    student.set_maths(row['Maths'])
    student.set_physics(row['Physics'])
    student.set_programming(row['Programming'])
    student.set_attendance(row['Attendance'])

    students.append(student)

for student in students:
    print(student)
    print("Scholarship Eligible:", student.scholarship_eligible())
    print(student.performance_warning())
    print("-" * 40)
```

## 12. Student Performance Report

```python
def student_report(df):
    print("----------- Student Report -----------")
    print()

    print("Total Students :", len(df))

    top_performer = df.loc[df['Total'].idxmax(), 'Name']
    print("Top Performer :", top_performer)

    highest_total = df['Total'].max()
    print("Highest Total Marks :", highest_total)

    passed = len(df[df['Result'] == 'Pass'])
    print("Students Passed :", passed)

    failed = len(df[df['Result'] == 'Fail'])
    print("Students Failed :", failed)

    print()
    print("--------------------------------------")

student_report(df)
```

## 13. Stack Implementation Using Array

```python
class Stack:
    def __init__(self):
        self.stack = []

    # Push element onto the stack
    def push(self, item):
        self.stack.append(item)
        print(f"{item} pushed into the stack.")

    # Pop element from the stack
    def pop(self):
        if len(self.stack) == 0:
            print("Stack Underflow")
        else:
            item = self.stack.pop()
            print(f"{item} popped from the stack.")

    # Display the stack
    def display(self):
        print("Stack:", self.stack)


# Create a stack object
s = Stack()

# Push elements
s.push(13)
s.push(13)
s.push(78)
s.push(67)

# Display stack
s.display()

# Pop two times
s.pop()
s.pop()

# Display stack after popping
s.display()
```

## 14. Stack Implementation Using A Linked List

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class Stack:
    def __init__(self):
        self.top = None

    # Push an element onto the stack
    def push(self, data):
        new_node = Node(data)
        new_node.next = self.top
        self.top = new_node

    # Pop an element from the stack
    def pop(self):
        if self.top is None:
            print("Stack Underflow")
        else:
            popped = self.top.data
            self.top = self.top.next
            print(f"{popped} popped from the stack.")

    # Display the stack
    def display(self):
        if self.top is None:
            print("Stack is empty")
        else:
            current = self.top
            print("Stack (Top to Bottom): ", end="")
            while current:
                print(current.data, end=" ")
                current = current.next
            print()


# Create a stack object
s = Stack()

# Push elements
s.push(13)
s.push(13)
s.push(78)
s.push(67)

# Display the stack
s.display()

# Pop two times
s.pop()
s.pop()

# Display the stack after popping
s.display()
```
