# Question 1: Temperature Analysis Using NumPy

```python
# import numpy, matplotlib, and scipy libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats, linalg, optimize, integrate
```

```python
temperatures = np.array([31, 29, 34, 28, 35, 32, 30, 36, 27, 33])

print("Average Temperature:", np.mean(temperatures))
print("Hottest Temperature:", np.max(temperatures))
print("Coldest Temperature:", np.min(temperatures))
print("Temperatures above average:", temperatures[temperatures > np.mean(temperatures)])
```

## Question 2: Branch-wise Sales Analysis

```python
sales = np.array([
    [120, 135, 128, 142],
    [150, 148, 160, 155],
    [110, 125, 118, 130]
])

print("Branch-wise total:")
print("Branch 1:", np.sum(sales[0]))
print("Branch 2:", np.sum(sales[1]))
print("Branch 3:", np.sum(sales[2]))
print()

print("Week-wise total:")
print("Week 1:", np.sum(sales[:,0]))
print("Week 2:", np.sum(sales[:,1]))
print("Week 3:", np.sum(sales[:,2]))
print("Week 4:", np.sum(sales[:,3]))
print()

highest = np.array([np.sum(sales[0]), np.sum(sales[1]), np.sum(sales[2])])
branches = np.array(["Branch 1", "Branch 2", "Branch 3"])

if np.max(highest) == highest[0]:
    print("Highest Selling Branch:", branches[0])
elif np.max(highest) == highest[1]:
    print("Highest Selling Branch:", branches[1])
else:
    print("Highest Selling Branch:", branches[2])

highest_week = np.array([np.sum(sales[:,0]), np.sum(sales[:,1]), np.sum(sales[:,2]), np.sum(sales[:,3])])
weeks = np.array(["Week 1", "Week 2", "Week 3", "Week 4"])

if np.max(highest_week) == highest_week[0]:
    print("Highest selling week:", weeks[0])
elif np.max(highest_week) == highest_week[1]:
    print("Highest selling week:", weeks[1])
elif np.max(highest_week) == highest_week[2]:
    print("Highest selling week:", weeks[2])
else:
    print("Highest selling week:", weeks[3])
```

## Question 3: Employee Sales File Processing

```python
with open("employees.txt", "w") as file:
    file.write("Alice 45000\n")
    file.write("Bob 32000\n")
    file.write("Charlie 58000\n")
    file.write("David 27000\n")
    file.write("Eva 51000\n")

with open("employees.txt", "r") as file:
    for line in file:
        print(line.strip())
```

```python
monthly_sales = []
names_all = []

with open("employees.txt", "r") as file:
    for line in file:
        names, sales = line.split(" ")
        sales = int((sales))
        monthly_sales.append((sales))
        names_all.append(names)

names = np.array(names_all)
sales = np.array(monthly_sales)
print("Total sales:", np.sum(sales))
print("Average sales:", np.mean(sales))
```

```python
filtered_sales = sales[sales > np.mean(sales)]
print(filtered_sales)
filtered_names = []
for i in monthly_sales:
    if i > np.mean(i):
        filtered_names.append(names_all[i])
print(filtered_names)
```

## Question 4: Student Result Analysis from CSV

```python
marks = np.array([
    [101, 78, 82],
    [102, 65, 71],
    [103, 90, 88],
    [104, 72, 69],
    [105, 85, 94]
])

n, d = marks.shape

final_score = np.array([
    [101, 0],
    [102, 0],
    [103, 0],
    [104, 0],
    [105, 0]
])

for i in range(0, n):
    fs = 0.40 * marks[i, 1] + 0.60 * marks[i, 2]
    final_score[i, 1] = fs

scores = final_score[final_score[:, 1] >= 75]

print("Student\t", "Final Score\t")
x, y = scores.shape

for i in range(0, x):
    print(f"S{scores[i, 0]}\t {scores[i, 1]}")
    if scores[i, 1] == np.max(scores[:,1:]):
        print(f"Highest scorer: S{scores[i, 0]}")
```

## Question 5: Delivery Time Statistical Analysis

```python
times = np.array([32, 45, 28, 51, 37, 42, 29, 95, 34, 41])

print("Mean:", np.mean(times))
print("Median:", np.median(times))
print("Mode:", stats.mode(times).mode)
print("Highest z-score value:", np.max(stats.zscore(times)))
```

## Question 7: Solve a Business Equation System

```python
A = np.array([
    [1, 1],
    [50, 80]
])

b = np.array([80, 5600])

solution = linalg.solve(A, b)
x = solution[0]
y = solution[1]

print("Product A units:", x)
print("Product B units:", y)
print()
print("Equation 1 verified:", x + y == 80)
print("Equation 1 verified:", 50 * x + 80 * y == 5600)
```
