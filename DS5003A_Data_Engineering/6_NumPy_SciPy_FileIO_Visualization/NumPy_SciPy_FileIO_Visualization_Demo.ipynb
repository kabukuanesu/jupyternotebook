## 1. Import Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats, linalg, optimize, integrate

print("NumPy version:", np.__version__)
```

# NumPy

NumPy is used for efficient numerical computation using arrays.

## 2. Creating NumPy Arrays

```python
a = np.array([10, 20, 30, 40, 50])
b = np.array([[1, 2, 3],
              [4, 5, 6]])

print(a)
print(b)
```

## 3. Array Properties

```python
print("Shape:", b.shape)
print("Dimensions:", b.ndim)
print("Number of elements:", b.size)
print("Data type:", b.dtype)
```

## 4. Special Arrays

```python
print("Zeros:")
print(np.zeros((2, 3)))

print("Ones:")
print(np.ones((2, 2)))

print("Range:")
print(np.arange(0, 10, 2))

print("Evenly spaced values:")
print(np.linspace(0, 1, 5))
```

## 5. Indexing and Slicing

```python
numbers = np.array([5, 10, 15, 20, 25, 30])

print(numbers[0])
print(numbers[-1])
print(numbers[1:4])
print(numbers[::2])
```

## 6. Reshaping Arrays

```python
x = np.arange(1, 13)
print("Original:", x)

matrix = x.reshape(3, 4)
print("Reshaped array:")
print(matrix)
```

## 7. Arithmetic Operations

```python
x = np.array([10, 20, 30])
y = np.array([1, 2, 3])

print("Addition:", x + y)
print("Subtraction:", x - y)
print("Multiplication:", x * y)
print("Division:", x / y)
print("Square:", x ** 2)
```

## 8. Aggregate Functions

```python
marks = np.array([72, 85, 91, 68, 88])

print("Sum:", np.sum(marks))
print("Mean:", np.mean(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Standard deviation:", np.std(marks))
```

## 9. Boolean Filtering

```python
marks = np.array([45, 67, 82, 39, 91, 76])

print(marks > 60)
print("Marks greater than 60:", marks[marks > 60])
```

## 10. Matrix Operations

```python
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])

print("Element-wise multiplication:")
print(A * B)

print("Matrix multiplication:")
print(A @ B)

print("Transpose:")
print(A.T)
```

# File I/O

Python can read data from files and write results to files.

## 11. Writing to a Text File

```python
with open("students_demo.txt", "w") as file:
    file.write("Anu 82\n")
    file.write("Binu 75\n")
    file.write("Cathy 91\n")

print("students_demo.txt created")
```

## 12. Reading a Text File

```python
with open("students_demo.txt", "r") as file:
    content = file.read()

print(content)
```

## 13. Reading a File Line by Line

```python
with open("students_demo.txt", "r") as file:
    for line in file:
        print(line.strip())
```

## 14. Appending to a File

```python
with open("students_demo.txt", "a") as file:
    file.write("Deepak 88\n")

with open("students_demo.txt", "r") as file:
    print(file.read())
```

## 15. Writing and Reading Numeric Data with NumPy

```python
scores = np.array([[101, 78, 82],
                   [102, 88, 91],
                   [103, 69, 74],
                   [104, 95, 90]])

np.savetxt("scores_demo.csv", scores, delimiter=",", fmt="%d")

loaded_scores = np.loadtxt("scores_demo.csv", delimiter=",", dtype=int)
print(loaded_scores)
```

## 16. Extracting Columns from File Data

```python
roll_no = loaded_scores[:, 0]
test1 = loaded_scores[:, 1]
test2 = loaded_scores[:, 2]

print("Roll numbers:", roll_no)
print("Test 1 marks:", test1)
print("Test 2 marks:", test2)
```

# SciPy

SciPy provides scientific computing functions built on top of NumPy.

## 17. Basic Statistics using scipy.stats

```python
data = np.array([12, 15, 18, 18, 20, 22, 25])

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Mode:", stats.mode(data).mode)
print("Z-scores:", stats.zscore(data))
```

## 18. Pearson Correlation

```python
study_hours = np.array([1, 2, 3, 4, 5, 6])
marks = np.array([45, 50, 58, 65, 72, 80])

r, p = stats.pearsonr(study_hours, marks)
print("Correlation coefficient:", r)
print("p-value:", p)
```

## 19. Solving Linear Equations with scipy.linalg

Solve:

$2x + y = 5$  
$x + 3y = 6$

```python
A = np.array([[2, 1],
              [1, 3]])
b = np.array([5, 6])

solution = linalg.solve(A, b)
print("x =", solution[0])
print("y =", solution[1])
```

## 20. Finding a Root with scipy.optimize

```python
def f(x):
    return x**2 - 9

root = optimize.root_scalar(f, bracket=[0, 5])
print("Root:", root.root)
```

## 21. Numerical Integration

Calculate the area under $f(x)=x^2$ from 0 to 3.

```python
def f(x):
    return x**2

area, error = integrate.quad(f, 0, 3)
print("Area:", area)
print("Estimated error:", error)
```

# Visualization with Matplotlib

Matplotlib is used to create graphs and charts.

## 22. Line Plot

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 18, 25, 35, 48])

plt.plot(x, y, marker="o")
plt.xlabel("Day")
plt.ylabel("Sales")
plt.title("Daily Sales")
plt.grid(True)
plt.show()
```

## 23. Bar Chart

```python
subjects = ["Maths", "Physics", "Programming", "English"]
marks = [82, 76, 91, 70]

plt.bar(subjects, marks)
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.title("Marks by Subject")
plt.show()
```

## 24. Scatter Plot

```python
hours = np.array([1, 2, 3, 4, 5, 6])
marks = np.array([42, 51, 59, 63, 75, 84])

plt.scatter(hours, marks)
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

## 25. Histogram

```python
marks = np.array([45, 52, 56, 61, 64, 67, 70, 72, 75, 78, 82, 85, 88, 91, 95])

plt.hist(marks, bins=5, edgecolor="black")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Distribution of Marks")
plt.show()
```

## 26. Pie Chart

```python
activities = ["Study", "Sleep", "Class", "Other"]
hours = [6, 8, 5, 5]

plt.pie(hours, labels=activities, autopct="%1.1f%%")
plt.title("Daily Time Distribution")
plt.show()
```

## 27. Multiple Lines in One Plot

```python
days = np.array([1, 2, 3, 4, 5])
student1 = np.array([60, 65, 70, 74, 80])
student2 = np.array([55, 63, 69, 78, 85])

plt.plot(days, student1, marker="o", label="Student 1")
plt.plot(days, student2, marker="s", label="Student 2")
plt.xlabel("Test")
plt.ylabel("Score")
plt.title("Score Comparison")
plt.legend()
plt.grid(True)
plt.show()
```

## 28. Saving a Plot to a File

```python
x = np.arange(1, 6)
y = x ** 2

plt.plot(x, y, marker="o")
plt.xlabel("x")
plt.ylabel("x squared")
plt.title("Square Values")
plt.savefig("square_plot.png", dpi=150, bbox_inches="tight")
plt.show()

print("square_plot.png saved")
```

## 29. Combined Example: File I/O + NumPy + SciPy + Visualization

The following example reads test scores from the CSV file created earlier, computes statistics, checks the relationship between two tests, and plots the scores.

```python
data = np.loadtxt("scores_demo.csv", delimiter=",", dtype=int)

roll = data[:, 0]
test1 = data[:, 1]
test2 = data[:, 2]

print("Average Test 1 mark:", np.mean(test1))
print("Average Test 2 mark:", np.mean(test2))

r, p = stats.pearsonr(test1, test2)
print("Correlation between Test 1 and Test 2:", r)

plt.plot(roll, test1, marker="o", label="Test 1")
plt.plot(roll, test2, marker="s", label="Test 2")
plt.xlabel("Roll Number")
plt.ylabel("Marks")
plt.title("Student Test Scores")
plt.legend()
plt.grid(True)
plt.show()
```
