```python
data = np.array([89, 68, 78, 96, 21, 45, 45, 56])

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Mode:", stats.mode(data).mode)
print("Z-Scores:", stats.zscore(data))
```

```python
study_hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])
marks = np.array([56, 65, 70, 71, 76, 80, 85, 91])

r, p = stats.pearsonr(study_hours, marks)
print("Correlation Coefficient:", r)
print("p-value", p)
```

```python
# Solve
# 2x +  y = 5
#  x + 3y = 6

A = np.array([
    [2, 1],
    [1, 3]
])

b = np.array([5, 6])

solution = linalg.solve(A, b)
print("Solution:", solution)
print("x =", solution[0])
print("y =", solution[1])
```

```python
def f(x):
    return x**2 -9

root = optimize.root_scalar(f, bracket=[0, 5])
print("Root:", root.root)
```

```python
def f(x):
    return x**2

area, error = integrate.quad(f, 0, 3)
print("Area:", area)
print("Estimated Error:", error)
```

```python
days = np.array([1, 2, 3, 4, 5])
student1 = np.array([60, 65, 70, 74, 80])
student2 = np.array([55, 63, 69, 78, 85])

plt.plot(days, student1, marker="o", label="Student 1")
plt.plot(days, student2, marker="s", label="Student 2")
plt.xlabel("Test")
plt.ylabel("Score")
plt.title("Score Comparison")
plt.legend()
plt.grid(True)
plt.show()
```

```python
data = np.loadtxt("score_demo.csv", delimiter=",", dtype=int)

roll = data[:, 0]
test1 = data[:, 1]
test2 = data[:, 2]

print("Average Test 1 mark:", np.mean(test1))
print("Average Test 2 mark:", np.mean(test2))

r, p = stats.pearsonr(test1, test2)
print("Correlation between Test 1 and Test 2:", r)

plt.plot(roll, test1, marker="o", label="Test 1")
plt.plot(roll, test2, marker="s", label="Test 2")
plt.xlabel("Roll Number")
plt.ylabel("Marks")
plt.title("Student Test Scores")
plt.legend()
plt.grid(True)
plt.show()
```
